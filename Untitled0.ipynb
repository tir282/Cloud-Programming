TCS CODE 2  fraud detection


```python
n = int(input("Enter n: "))
d={}
for _ in range(n):
  s=input("Enter String: ").split()
  sender= s[0]
  receiver= s[1]
  amount=int(s[2])
  time=int(s[3])
  key=(sender,receiver,amount)

  # The original code had print(d[key]) here, which caused KeyError
  # because d is initially empty. This line has been removed.

  if key in d and abs(time -d[key])<60 :
    print("FRAUD: " ,sender,receiver,amount,time)

  d[key]=time
  print(f"Key stored in dictionary: {key}")
  print(f"Current time value: {time}")
```

TCS CODE 3 Parking fee cancelation


```python
def exclusiveTime(n, logs):
    result = [0] * n
    stack = []
    prev_time = 0

    for log in logs:
        fn_id, typ, time = log.split(':')
        fn_id = int(fn_id)
        time = int(time)

        if typ == "start":
            if stack:
                result[stack[-1]] += time - prev_time
            stack.append(fn_id)
            prev_time = time

        else:  # "end"
            result[stack.pop()] += time - prev_time + 1
            prev_time = time + 1

    return result


# -------- USER INPUT --------
n = int(input())          # number of functions
m = int(input())          # number of logs

logs = []
for _ in range(m):
    logs.append(input().strip())

# -------- OUTPUT --------
print(exclusiveTime(n, logs))
```

```python
def factorial_value(num):

  fact = 1
  if num == 0:
    return fact

  else:
    for i in range(1, num+1):
      fact = fact*i
    return fact
```

```python
print(factorial_value(10))
```
